```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import normaltest

numeric_features = df.select_dtypes(include=np.number).columns

# Plot histograms & density, and then 3. Perform normality test
for col in numeric_features:
    plt.figure(figsize=(8, 4))

    # Plot histogram with KDE
    sns.histplot(data=df, x=col, kde=True, bins=20)
    plt.title(f'{col} Distribution', fontsize=14)
    plt.xlabel(col, fontsize=12)
    plt.ylabel('Frequency', fontsize=12)
    plt.tight_layout()
    plt.show()

    stat, pval = normaltest(df[col].dropna())  # dropna in case of missing values
    if pval < 0.05:
        print(f'{col}: Not Gaussian (p-value = {pval:.5f}, stat = {stat:.5f})')
    else:
        print(f'{col}: Gaussian (p-value = {pval:.5f}, stat = {stat:.5f})')
    print('-'*60)
```

```python
numeric_cols = df.select_dtypes(include=[np.number]).columns
print("Numeric columns:", numeric_cols)

outliers_dict = {}

for col in numeric_cols:
    # Drop NaN values if they exist in this column
    col_data = df[col].dropna()

    # Calculate Q1, Q3, and IQR
    Q1 = col_data.quantile(0.25)
    Q3 = col_data.quantile(0.75)
    IQR = Q3 - Q1

    # Define outlier cutoffs
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Identify outliers
    is_outlier = (df[col] < lower_bound) | (df[col] > upper_bound)
    outliers = df[is_outlier]

    # Store outliers in a dictionary if you want to reference later
    outliers_dict[col] = outliers

    # Print summary
    print(f"\nColumn: {col}")
    print(f"  Q1 = {Q1}, Q3 = {Q3}, IQR = {IQR}")
    print(f"  Lower Bound = {lower_bound}, Upper Bound = {upper_bound}")
    print(f"  Number of outliers = {outliers.shape[0]}")
```

## Handeling Missing Values

```python
#missing value percentage
missing_percentage = (df.isnull().sum() / len(df)) * 100
missing_percentage.sort_values(ascending=False)
```

```python
#plot a heatmeap for missing values
plt.figure(figsize=(10, 6))
sns.heatmap(df.isnull(), cbar=False, cmap='viridis')
plt.title('Missing Values Heatmap')
plt.show()
```

```python
#plot a bar chart for missing value percetange
plt.figure(figsize=(10, 6))
missing_percentage.plot(kind='bar')
plt.title('Missing Value Percentage')
plt.xlabel('Columns')
plt.ylabel('Percentage')
plt.xticks(rotation=45)
plt.show()
```
